# Data Cleaning
## Find and Remove columns with 80 % missing values

The data is checked for Null values in each column using .info(), since missing values introduce biased and invalid results. Large number of 0 or Null values will be given more weight hence, the columns with more than 80 % missing values are removed.

```python
#sum of Null values in each column
miss_train = df.isnull().sum().sort_values(ascending=False)
#find column index with 80 % missing values
train_mask = (miss_train)/2930 > 0.20
#index the columns with large no. of missing values
miss_train[train_mask].index
```

'Pool QC', 'Misc Feature', 'Alley', 'Fence', 'Fireplace Qu' are the columns with large null values and hence can be dropped from further analysis since, they may not add much contibution to House price prediciton but would rather introduce biased and inaccurate results. The values can be filled by replacing with the average or dummy values however, after looking at the description of the attributes it is concluded that they do not hold much importance while selecting the House Price. Pool Quality, Miscellaneous feature not covered in other categories, Type of alley access to property, Fence quality and  Fireplace quality are minor features that donot usually add much to the House Price. <br>
But before dropping the 5 columns, univariate analysis is performed to further strengthen the reasoning.

### Bar Graph for Categorical Attributes

```python
# Plot bar plot for each categorical feature
fig, axes =plt.subplots(10,4, figsize=(20,40), sharex=True)
axes = axes.flatten()
object_bol = df.dtypes == 'object'
for ax, catplot in zip(axes, df.dtypes[object_bol].index):
    sns.countplot(y=catplot, data=df, ax=ax)
plt.tight_layout()  
plt.show()
```

## Bivariate Analysis
### Heat Map

```python
#HeatMap for multivariate analysis
plt.figure(figsize = (35,20))
cmap = sns.diverging_palette(230,20,as_cmap=True)
sns.heatmap(df.corr(),annot=True,fmt='.2f',cmap=cmap)
plt.show()
```

The HeatMap shows the Bivariate Analysis with dark color shades of Red showing high correlation between attributes and dark color shades of Blue showing high Negative correlation between attributes. 

*   SalesPrice is positively correlated with Overall Qual, Gr Liv Area
*   Garage Yr Blt is positively correlated with Year Built
*   Gr Liv Area is positively correlated with Tot Rms AbvGrd
*   Garage Cars is positively correlated with Garage Area<br>

These are the values with correlation greater than or equal to 0.81. <br><br>

*   Bsmt Unf SF is negatively correlated with BsmtFin SF
*   BsmtFin SF 1 is negatively correlated with Bsmt Unf SF <br>

These are the values with correlation less than or equal to -0.48. <br><br>

The attributes decided to be removed after Univariate analysis using Bar Graph do not show any significant correlation in the HeatMap as well and hence, provides another reason to remove them. The attributes were: Street, Land Contour, Utilities, Land Slope, Condition 1 , Condition 2, Roof Mtl, Heating, BsmtFin Type 2, Functional and Garage Cond. They are removed below.
