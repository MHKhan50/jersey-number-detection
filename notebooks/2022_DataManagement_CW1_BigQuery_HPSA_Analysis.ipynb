````python
from google.cloud import bigquery

client = bigquery.Client(project='data-management-landscape')
````

````python
df = client.query('''
SELECT COUNT(Source_Name), Status_Description, Discipline_Class_Desc From
(SELECT Source_Name, Status_Description, Discipline_Class_Desc  From ((
SELECT Source_Name, Status_Description,Discipline_Class_Desc  From bigquery-public-data.sdoh_hrsa_shortage_areas.hpsa_primary_care)
UNION ALL
(SELECT Source_Name, Status_Description, Discipline_Class_Desc From bigquery-public-data.sdoh_hrsa_shortage_areas.hpsa_mental_health)
UNION ALL
(SELECT Source_Name, Status_Description,Discipline_Class_Desc From bigquery-public-data.sdoh_hrsa_shortage_areas.hpsa_dental_health
) ) )
GROUP BY Status_Description, Discipline_Class_Desc;

''').to_dataframe()
df
````

````python
df.to_csv("question1_query_result.csv", index=False)
````

#Answer to Question 2

````python
df = []
subdata = []
df = client.query('''
SELECT HPSA_Score, State_Name, Discipline_Class_Desc From ((
SELECT HPSA_Score, State_Name,Discipline_Class_Desc  From bigquery-public-data.sdoh_hrsa_shortage_areas.hpsa_primary_care)
UNION ALL
(SELECT HPSA_Score, State_Name,Discipline_Class_Desc From bigquery-public-data.sdoh_hrsa_shortage_areas.hpsa_mental_health)
UNION ALL
(SELECT HPSA_Score, State_Name,Discipline_Class_Desc From bigquery-public-data.sdoh_hrsa_shortage_areas.hpsa_dental_health
) ) WHERE HPSA_Score IS NOT NULL
''').to_dataframe()
df
````

````python
subdata = df.copy()
````

````python
subdata = df[['State_Name', 'HPSA_Score','Discipline_Class_Desc']]
subdata = subdata.groupby(['State_Name', 'Discipline_Class_Desc']).mean().round().reset_index(drop=False)
# subdata = subdata.groupby(['State_Name']).count().reset_index(drop=False)
# subdata = subdata.reset_index(drop=False)
# subdata = subdata[[col1,col2,'']]
subdata.columns = ['State_Name','Class','HPSA_Score']
subdata
````

````python
df.to_csv("question2_query_result.csv", index=False)
````

````python
fig1 = px.bar(subdata, x='State_Name', y='HPSA_Score',color='Class', title="HPSA Score for areas of various classes")
fig1
````
