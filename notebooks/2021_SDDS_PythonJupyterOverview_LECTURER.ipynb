**Functions**


```
# this function that takes 2 numbers as input and outputs their average value
def avg(x,y):
	print("first input is ",  x )
	print("second input is ", y)
	a = (x + y)/2.0
	print("average is ", a)
	return a

#call the function
avg(200,300)
```

**Arrays**


```
import numpy
#the import command tells python to bring in additional functionality from elsewhere
#here we have imported an additional set of tools packaged into a module called
#numpy

a = numpy.zeros([3,2])
print(a)
a[0,0]  =  1
a[0,1 ] =  2
a[1,0 ] =  9
a[2,1] = 12
print(a)
```

```
#note that python arrays are indexed from 0 and not 1
print(a[0,1])
v = a[1,0]
print(v)
```
